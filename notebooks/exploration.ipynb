# Exploration du dataset Churn

Analyse exploratoire du jeu de données synthétique de churn télécom. Ce notebook est hors pipeline de production : tout ce qui est transformé ici doit être répliqué dans `src/features/build_features.py` (pas de training/serving skew).

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("data/raw/dataset.csv")
print(df.shape)
df.head()
```

```python
df.describe(include="all").T
```

```python
print(df.isna().sum()[df.isna().sum() > 0])
print(f"churn rate: {df['Churn'].value_counts(normalize=True).to_dict()}")
```

```python
df.groupby("Contract")["Churn"].value_counts(normalize=True).unstack().plot(kind="bar", stacked=True)
plt.title("Taux de churn par type de contrat")
plt.show()
```

```python
df.groupby("InternetService")["Churn"].value_counts(normalize=True).unstack().plot(kind="bar", stacked=True)
plt.title("Taux de churn par type de service internet")
plt.show()
```

```python
df["Tenure"].hist(bins=30)
plt.title("Distribution de l'ancienneté (Tenure)")
plt.show()
```

## Conclusion

- Pas de valeurs manquantes sur les colonnes clés.
- Le churn est corrélé au type de contrat (month-to-month) et au service internet (fibre).
- Les transformations retenues (encodage des catégories, features dérivées `charges_per_tenure`, `tenure_years`, `num_services`) sont encapsulées dans `FeatureTransformer`.
